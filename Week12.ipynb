Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Museum visitors for each day
week1 = [100, 110, 130, 120, 100, 90, 80]
week2 = [70, 90, 100, 100, 120, 110, 90]

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

# Create two separate graphs
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

axs[0].plot(days, week1, marker='o', label='Week 1')
axs[0].set_title('Museum Visitors - Week 1')
axs[0].set_ylabel('Number of Visitors')
axs[0].tick_params(axis='x', rotation=45)
axs[0].grid(True)

axs[1].plot(days, week2, marker='o', color='orange', label='Week 2')
axs[1].set_title('Museum Visitors - Week 2')
axs[1].set_ylabel('Number of Visitors')
axs[1].tick_params(axis='x', rotation=45)
axs[1].grid(True)

plt.tight_layout()
plt.show()

# Combine both weeks in one graph
plt.figure(figsize=(10, 5))

plt.plot(days, week1, marker='o', label='Week 1')
plt.plot(days, week2, marker='o', label='Week 2')

plt.title('Comparison of Museum Visitors: Week 1 vs. Week 2')
plt.xlabel('Day')
plt.ylabel('Number of Visitors')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Presenting the museum visitor data in two separate graphs makes it easier to focus on the daily visitor patterns within each week. Each graph is less crowded, allowing readers to examine changes in visitor numbers independently. However, comparing the two weeks directly can be challenging because readers must look back and forth between the graphs.

Another issue with using separate graphs is that Matplotlib automatically creates a different y-axis scale for each graph by default. For example, one graph may show a range of approximately 80 to 130 visitors, while the other shows approximately 70 to 120. Because the scales are different, the same visual height does not represent the same number of visitors. This can make the comparison misleading. Using the same y-axis scale for both graphs with `sharey=True` would make the comparison more accurate.

Presenting both weeks on the same graph makes comparison easier because the visitor numbers for each day can be viewed together. This approach helps identify differences and similarities between the two weeks and makes overall patterns more noticeable. Using different colors and a clear legend can help distinguish between the weeks. However, multiple lines may make the graph confusing if too many weeks are included.

Overall, I would prefer displaying both weeks on the same graph because the main purpose is to compare visitor numbers across the two weeks. A combined graph with clearly labeled lines and a shared y-axis makes the differences easier to understand and provides a more accurate visual comparison.

```python
import matplotlib.pyplot as plt
import numpy as np

products = ['Product A', 'Product B', 'Product C',
            'Product D', 'Product E']

years = ['2008', '2009', '2010', '2011',
         '2012', '2013', '2014']

data = {
    '2008': [390, 360, 0, 0, 0],
    '2009': [420, 400, 0, 0, 0],
    '2010': [425, 410, 100, 0, 0],
    '2011': [385, 375, 180, 160, 0],
    '2012': [300, 290, 200, 260, 0],
    '2013': [270, 260, 240, 220, 100],
    '2014': [260, 250, 180, 215, 210]
}

x = np.arange(len(products))
width = 0.11

fig, ax = plt.subplots(figsize=(12, 6))

for i, year in enumerate(years):
    offset = (i - 3) * width
    ax.bar(x + offset, data[year], width, label=year)

ax.set_title(
    'Price Has Declined for Products A and B Since 2010',
    fontsize=16,
    pad=20
)

ax.set_ylabel('Average Retail Price ($)')
ax.set_xlabel('Products')
ax.set_xticks(x)
ax.set_xticklabels(products)

ax.legend(title='Year', frameon=False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

For this assignment, I recreated a grouped bar chart in the style of an example from Chapter 8 of *Storytelling with Data*. The graph compares the average retail prices of five products from 2008 to 2014. The title states the main message: the prices of Products A and B have declined since 2010, while newer products entered the market. I grouped the bars by product, used a different color for each year, removed the top and right borders, and kept the labels clear to reduce clutter. This shows how a chart can communicate a message instead of only displaying numbers.
